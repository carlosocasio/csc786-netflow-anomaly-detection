```python
import pandas as pd
import numpy as np
import os

# Use a fixed seed so results are reproducible
rng = np.random.default_rng(42)

# Generate 1,000 synthetic NetFlow records
num_records = 1000

data = {
    "timestamp": pd.date_range(
        start="2026-10-01 08:00:00",
        periods=num_records,
        freq="10s"
    ),
    "src_ip": [
        f"10.10.10.{i}"
        for i in rng.integers(1, 255, num_records)
    ],
    "dst_ip": [
        f"10.20.20.{i}"
        for i in rng.integers(1, 255, num_records)
    ],
    "src_port": rng.integers(1024, 65536, num_records),
    "dst_port": rng.choice([22, 53, 80, 443, 8080], num_records),
    "protocol": rng.choice(["TCP", "UDP"], num_records),
    "packets": rng.integers(1, 1000, num_records),
    "bytes": rng.integers(100, 100000, num_records)
}

df = pd.DataFrame(data)

print("Total NetFlow records:", len(df))
display(df.head(1000))
```

```python
os.makedirs("data", exist_ok=True)
df.to_csv("data/netflow_sample.csv", index=False)

print("NetFlow data saved successfully!")
```
